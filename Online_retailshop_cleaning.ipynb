```python
## Drop columns where there is no customer id
## in pandas dropna drops columns where the value is null( Nan)
retail_df1 = retail_df.dropna(subset = ['CustomerID'])
retail_df1.isnull().sum()




                                                        
                                                        
                                    
                                                        
```

```python
## show the values that are duplicated
retail_df1.duplicated()
retail_df1[retail_df1.duplicated()]
```

```python
## Keep the first occurence of duplicates, delete the rest
retail_df1= retail_df1.drop_duplicates()
```

```python
retail_df1.duplicated().sum()
```

```python
retail_df1.shape
```

```python
retail_df1.info()
```

```python
## Transform the CustomerID data type
retail_df1['CustomerID'] = retail_df1['CustomerID'].astype(int)
```

```python
## check if the Country names have been standardized
countries= retail_df1['Country'].unique()
countries
```

```python
## Create a filter to remove rows with -tive value for quantity and 0 for unit price
retail_df_clean = retail_df1[(retail_df1['Quantity']>= 1) & (retail_df1['UnitPrice']> 0)].copy()
```

```python
retail_df_clean.shape
```

```python
## check if the description names have been standardized
desc= retail_df1['Description'].unique()
desc
```

```python
### Run a summary statistics of the cleaned dataset
retail_df_clean[['InvoiceDate', 'Quantity', 'UnitPrice']].describe()
```

```python
## Create the revenue column
retail_df_clean['Revenue']= retail_df_clean['Quantity'] * retail_df_clean['UnitPrice']
retail_df_clean.head()
```

```python
### Run a summary statistics of the cleaned dataset
retail_df_clean[['InvoiceDate', 'Quantity', 'UnitPrice','Revenue']].describe()
```

#### Rows: 392,692 (Cleaned and verified).

#### Columns: All original columns plus the new Revenue column.

#### Quality: No negatives, no zeroes, and no missing customer identifiers.
